# バイブコーディング制作回 — 自分のアイデアを形にする

この回は、生成AIと一緒に自分のアイデアでプログラムを作る回です（3時間×2回）。

- 第1回: 環境セットアップ → バイブコーディングの練習 → 制作開始（動く最小版まで）
- 第2回: 機能追加・仕上げ → 発表会

これまでの回で使った道具（`requests`・サーバ・`Gradio`・ハッシュ/暗号）は全部素材になります。

---
# バイブコーディングとは

日本語でやりたいことを伝えて、コードはAIに書かせる作り方です。ただし丸投げとは違います。この回では次のサイクルを守ります。

1. 仕様を書く — 何を作るかを先に日本語で3行で書く
2. AIに書かせる — 仕様を渡してコードを作らせる
3. 読む — 出てきたコードを読み、何をしているか説明できるか確かめる
4. 直す — 動かないときはエラーを自分で読んでから、AIに具体的に指示する

発表では「AIが書いたコードのどこを自分で直したか」を必ず1つ話してもらいます。読めない・直せないコードは自分の作品と言えないからです。

# セットアップ：Antigravity をインストールする

Antigravity は Google のAI開発環境です。エージェント（AI）が計画を立て、ファイルを作り、実行して確認するところまでやってくれます。無料で、Googleアカウントがあれば使えます。

## 動作環境

- Windows 10 / 11（64bit）
- Googleアカウント（ログインに必要）

## 手順

1. https://antigravity.google/download を開く
2. Download for x64 を押してダウンロードする
3. ダウンロードした Antigravity.exe を実行してインストールする（設定はそのままでOK）
4. Antigravity を起動して、Googleアカウントでログインする
5. テーマなどの初期設定を済ませる（あとから変えられます）

## Python が入っているか確認する

Antigravity のエージェントは、あなたのPCの Python を使ってコードを実行します。ターミナル（Antigravity 内のターミナルでOK）で次を実行して確認します。

```
python --version
```

`Python 3.x.x` と表示されれば準備完了です。表示されない場合は https://www.python.org/downloads/ からインストールしてください（インストール時に Add python.exe to PATH に必ずチェック）。

## 動作確認：エージェントに1本作らせる

1. Antigravity で新しいプロジェクト（空のフォルダ）を開く
2. チャット欄に次のように入力する

```
数当てゲームの guess.py を作って実行してください。
1〜100の乱数を当てるゲームで、大きい・小さいのヒントを出します。
```

3. エージェントがファイルを作り、実行するまでを見届ける
4. できた `guess.py` を開いて、自分で読んでみる。`random` や `while` や `if` は、これまでの回で全部習ったものです

ここまで動けばセットアップ完了です。

## インストールできない場合

Colab で進めます。Colab の Gemini（サイドパネル）に同じように日本語で指示すれば、この回の課題はすべて Colab でもできます。

# ウォームアップ：AIのコードを直す練習

制作に入る前に、サイクルの「読む」「直す」を練習します。

1. エージェントに次を頼む

```
おみくじプログラム omikuji.py を作ってください。
実行すると 大吉・中吉・小吉・凶 のどれかをランダムに表示します。
```

2. できたコードを読んで、各行が何をしているか隣の人に説明する
3. 自分で1箇所改造する（AIに頼まず手で直す）。例:
   - 運勢の種類を増やす
   - 大吉が出にくいように確率を変える
   - 名前を入力させて「〇〇さんの運勢は…」と表示する
4. 改造がうまく動かなかったら、エラーを読んでから AI に「この行でこういうエラーが出た」と具体的に相談する

# 制作メニュー

自由題でもいいですが、迷ったらここから選んでください。どれもこれまでの回の道具で作れます。

## 1. Gradio ミニアプリ（易）

診断メーカー、単位換算、時間割リマインダーなど。ミニハッキングゲーム回で使った `Gradio` で画面を作る。

最初のプロンプト例:
```
gradioで動く性格診断アプリを作ってください。
3つの質問に答えると診断結果が表示されます。python app.py で起動できるようにしてください。
```

## 2. 暗号メッセンジャー（中）

バイナリ回のXOR暗号とハッシュ署名を使って、「盗聴されても読めない・改ざんされたら分かる」メッセージ交換ツールを作る。

最初のプロンプト例:
```
XOR暗号で文章を暗号化・復号するツールを作ってください。
暗号文にはsha256の署名を付けて、改ざんされたら検知できるようにします。
鍵はプログラム内の変数で構いません。
```

## 3. 自作CTF問題（中）

CTF回で解いた側から、出題側に回る。base64・XOR・ハッシュを組み合わせた問題を作り、教室で交換して解き合う。

最初のプロンプト例:
```
pythonのCTF問題を1問作ってください。フラグ文字列をbase64とXOR暗号で二重に隠し、
解く人にはヒントだけを見せる problem.py と、答えの解説 solution.md に分けてください。
```

## 4. Webサービス公開（中〜難）

ネットワーク回の `http.server` で掲示板やランキングを作り、cloudflared で教室に公開する。

最初のプロンプト例:
```
pythonのhttp.serverで簡単な掲示板サーバを作ってください。
POST /post で書き込み、GET / で一覧がHTMLで見えるようにしてください。
```

## 5. アリーナ改造（難）

コイン争奪戦のサーバコードをAIに読ませて、新ルール（障害物・アイテム・鬼ごっこモード）を追加し、オリジナル対戦会を開く。

最初のプロンプト例:
```
このサーバコードを読んで、盤面に障害物を追加してください。
障害物のマスには移動できないようにします。まずコードの構造を説明してください。
```

# AI利用のルール

- 仕様（何を作りたいか）は自分のアイデアを出すこと。AIに丸投げしない。
- 出てきたコードは必ず読む。説明できないコードは使わない
- エラーが出たら、まず自分でエラーメッセージを読む。それからAIに「どの行で・何をしたら・何が出たか」を伝える
- 授業で習った道具（`requests`・`dict`・関数・ループなど）が出てきたら、これまでに習った知識を役立ちます。AIの書き方と自分の書き方を見比べてみよう
