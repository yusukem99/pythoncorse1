# クラスとは

クラス とは、オブジェクトを作り出すための 設計図 です。

これまで使ってきた `str`（文字列）や `list`（リスト）なども、実はすべてクラスから作られたオブジェクトです。

- `"tanaka"` … `str` クラスから作られたオブジェクト
- `[10, 20, 30]` … `list` クラスから作られたオブジェクト

クラスを使うと、自分だけのデータ型 を作れます。ここでは、カードゲームのプレイヤーを表す `Player` クラスを例に学んでいきます。

## クラスの定義（__init__ と self）

クラスは `class` キーワードで定義します。

- `__init__` は、インスタンス（オブジェクト）を作るときに自動で呼ばれる特別なメソッドです（初期化を行う）
- `__init__` の第1引数 `self` は、そのオブジェクト自身 を指します
- オブジェクトが持つ変数を プロパティ、オブジェクトが持つ関数を メソッド と呼びます

下の `Player` クラスは、名前（`name`）と持ちチップ数（`chips`）をプロパティに持ち、状態を表示する `show` メソッドを持ちます。

````python
class Player:
    def __init__(self, name, chips):
        # プロパティの設定（オブジェクト自身に値を持たせる）
        self.name = name
        self.chips = chips

    def show(self):
        # メソッド（オブジェクトが持つ関数）
        print("{}: {}チップ".format(self.name, self.chips))
````

## インスタンス化（設計図からオブジェクトを作る）

クラス（設計図）から実際のオブジェクトを作ることを インスタンス化 といい、作られたオブジェクトを インスタンス と呼びます。

````python
# Player クラスから2つのインスタンスを作る
alice = Player("alice", 100)
bob = Player("bob", 50)

# メソッドを呼び出す
alice.show()
bob.show()

# プロパティを参照する
print("aliceのチップ数:", alice.chips)
````

## 特殊メソッド（`>` で比べる）

クラスには `__` で始まる 特殊メソッド を定義できます。これを使うと、`>` や `<` などの演算子を自分のクラスでも使えるようになります（演算子のオーバーロード）。

- `__gt__` … `>` 演算子で呼ばれる（gt = greater than）
- `__lt__` … `<` 演算子で呼ばれる（lt = less than）

`alice > bob` と書くと、Pythonは自動的に `alice.__gt__(bob)` を呼び出します。このとき `self` が `alice`、`other` が `bob` になります。

下の `Player` クラスは、チップ数の大小 でプレイヤーを比較できるようにしています。

````python
class Player:
    def __init__(self, name, chips):
        self.name = name
        self.chips = chips

    def show(self):
        print("{}: {}チップ".format(self.name, self.chips))

    def __gt__(self, other):
        # self のチップ数が other より多いか
        return self.chips > other.chips

    def __lt__(self, other):
        # self のチップ数が other より少ないか
        return self.chips < other.chips


alice = Player("alice", 100)
bob = Player("bob", 50)

# チップ数で比較できる
print("alice は bob より多い？:", alice > bob)
print("alice は bob より少ない？:", alice < bob)
````
